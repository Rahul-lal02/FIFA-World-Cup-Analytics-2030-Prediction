```python
import math
import pandas as pd

# Convert datetime columns to native Python datetime
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True, errors="coerce")

# Convert every value to a native Python type
def sql_value(x):
    if pd.isna(x):
        return None
    return x.item() if hasattr(x, "item") else x

data = [
    tuple(sql_value(v) for v in row)
    for row in df.itertuples(index=False, name=None)
]

print("Prepared", len(data), "rows")
print(type(data[0][2]))   # home_score
print(type(data[0][3]))   # home_xg
print(type(data[0][16]))  # Date
```

```python
# Convert pandas Timestamp to native Python datetime
df["Date"] = df["Date"].dt.to_pydatetime()

# Rebuild the data tuples
def sql_value(x):
    if pd.isna(x):
        return None
    return x.item() if hasattr(x, "item") else x

data = [
    tuple(sql_value(v) for v in row)
    for row in df.itertuples(index=False, name=None)
]

print(type(data[0][16]))
```
